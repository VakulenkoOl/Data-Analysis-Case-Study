```python
#Знаходжу середнє, мінімальне та максимальне значення усіх змінних за допомогою .describe#

import pandas as pd

dataframe = pd.read_csv('dailyActivity_merged.csv')

dataframe.iloc[:, 1:].describe().round(1)
```

```python
#Знаходжу середні значення всіх змінних кожного дня впродовж тижня#

columns_to_avg = [
    'Calories', 'TotalSteps', 'TotalDistance', 
    'VeryActiveDistance', 'ModeratelyActiveDistance', 'LightActiveDistance', 
    'VeryActiveMinutes', 'FairlyActiveMinutes', 'LightlyActiveMinutes', 'SedentaryMinutes']

weekly_summary = dataframe.groupby('DayOfTheWeek')[columns_to_avg].mean().round(1)

days_order = ['понеділок', 'вівторок', 'середа', 'четвер', "пʼятниця", 'субота', 'неділя']

weekly_summary = weekly_summary.reindex(days_order)

weekly_summary
```

```python
#Знаходжу середні значення всіх змінних у кожного користувача#

columns_to_avg = [
    'Calories', 'TotalSteps', 'TotalDistance', 
    'VeryActiveDistance', 'ModeratelyActiveDistance', 'LightActiveDistance', 
    'VeryActiveMinutes', 'FairlyActiveMinutes', 'LightlyActiveMinutes', 'SedentaryMinutes']


users_summary = dataframe.groupby('Id')[columns_to_avg].mean().round(1)

users_summary
```

```python
#Зрупую користувачів за їхньою активністю в кроках#

user_avg = dataframe.groupby('Id')['TotalSteps'].mean().round(1).reset_index()
user_avg.columns = ['Id', 'AverageSteps']

def classify_activity(steps):
    if steps >= 10000:
        return 'High Activity'
    elif steps >= 5000:
        return 'Moderate Activity'
    else:
        return 'Low Activity'

user_avg['ActivityGroup'] = user_avg['AverageSteps'].apply(classify_activity)

user_avg = user_avg.sort_values(by='AverageSteps', ascending=False)

user_avg
```

```python
#Визначаю кількість користувачів у кожній групі та їхнє співвідношення#

user_avg = dataframe.groupby('Id')['TotalSteps'].mean().reset_index()
user_avg.columns = ['Id', 'AverageSteps']

def classify_activity(steps):
    if steps >= 10000:
        return 'High Activity'
    elif steps >= 5000:
        return 'Moderate Activity'
    else:
        return 'Low Activity'

user_avg['ActivityGroup'] = user_avg['AverageSteps'].apply(classify_activity)

activity_summary = user_avg['ActivityGroup'].value_counts().reset_index()
activity_summary.columns = ['ActivityGroup', 'UserCount']

activity_summary['Percentage'] = (activity_summary['UserCount'] / len(user_avg) * 100).round(1)

activity_summary
```

```python
#Визначаю середню, мінімальну та максимальну кількість калорій кожної години впродовж дня#

df_hourlyCalories = pd.read_csv('hourlyCalories_merged.csv')

hourlyCalories = df_hourlyCalories.groupby('ActivityHour')['Calories'].agg(['mean','min','max']).round(1)

hourlyCalories.columns = ['AvgCalories', 'MinCalories', 'MaxCalories']

hourlyCalories
```

```python
#Визначаю середню, мінімальну та максимальну кількість кроків кожної години впродовж дня#


df_hourlySteps = pd.read_csv('hourlySteps_merged.csv')

hourlySteps = df_hourlySteps.groupby('ActivityHour')['StepTotal'].agg(['mean', 'min','max']).round(1)

hourlySteps.columns = ['AvgSteps','MinSteps','MaxSteps']

hourlySteps
```

```python
#Визначаю середні час сну та час у ліжку, середню к-сть записів про сон, а також порахував ефективність сну впродовж тижня#

df_sleepDay = pd.read_csv('sleepDay_merged.csv')

df_sleepDay['SleepEfficiencyPercentage'] = (df_sleepDay['TotalMinutesAsleep'] / df_sleepDay['TotalTimeInBed'] * 100).round(1)

columns_to_avg = ['TotalMinutesAsleep','TotalTimeInBed','TotalSleepRecords','SleepEfficiencyPercentage']

weekly_summary = df_sleepDay.groupby('DayOfTheWeek')[columns_to_avg].mean().round(1)

days_order = ['понеділок', 'вівторок', 'середа', 'четвер', "пʼятниця", 'субота', 'неділя']

weekly_summary = weekly_summary.reindex(days_order)

weekly_summary
```

```python
#Розділив користувачів за ефективністю сну, визначив к-сть у кожній групі та їхнє співвідношення#

df_sleepDay = pd.read_csv('sleepDay_merged.csv')

user_sleep = df_sleepDay.groupby('Id')[['TotalMinutesAsleep', 'TotalTimeInBed']].sum().reset_index()

user_sleep['SleepEfficiencyPercentage'] = (user_sleep['TotalMinutesAsleep'] / user_sleep['TotalTimeInBed'] * 100).round(1)

def classify_SleepEfficiency(SleepEfficiency):
    if SleepEfficiency >= 85:
        return 'Good Sleep'
    elif SleepEfficiency >= 80:
        return 'Normal Sleep'
    else:
        return 'Bad Sleep'

user_sleep['SleepEfficiencyGroup'] = user_sleep['SleepEfficiencyPercentage'].apply(classify_SleepEfficiency)

sleep_summary = user_sleep['SleepEfficiencyGroup'].value_counts().reset_index()
sleep_summary.columns = ['SleepEfficiency', 'UserCount']

sleep_summary['Percentage'] = (sleep_summary['UserCount'] / len(user_sleep) * 100).round(1)

sleep_summary
```

```python
#Знаходжу середню інтенсивність користувачів за кожен день впродовж тижня#

df_intensities = pd.read_csv('hourlyIntensities_merged.csv')

weekly_intensity = df_intensities.groupby('DayOfTheWeek')['TotalIntensity'].mean().round(1)

days_order = ['понеділок', 'вівторок', 'середа', 'четвер', "пʼятниця", 'субота', 'неділя']

weekly_intensity = weekly_intensity.reindex(days_order)

weekly_intensity
```

```python
#Знаходжу середню інтенсивність користувачів кожної години впродовж дня#

hourly_intensity = df_intensities.groupby('ActivityHour')['TotalIntensity'].mean().round(1)

hourly_intensity
```

```python
#Розділив користувачів за їхнім ІМТ, визначив к-сть у кожній групі та їхнє співвідношення#

df_weight = pd.read_csv('weightLogInfo_merged.csv')

user_weight = df_weight.groupby('Id')['BMI'].mean().round(1).reset_index()

def bmi_classification(BMI):
    if BMI >= 30:
        return 'obesity'
    elif BMI >= 25:
        return 'excess weight'
    elif BMI >= 18.5:
        return 'normal weight'
    else:
        return 'underweight'

user_weight['WeightGroup'] = user_weight['BMI'].apply(bmi_classification)

weight_summary = user_weight['WeightGroup'].value_counts().reset_index()
weight_summary.columns = ['WeightGroup', 'UserCount']

weight_summary['Percentage'] = (weight_summary['UserCount'] / len(user_weight) * 100).round(1)

weight_summary
```
